```python
import pandas as pd

file_path = "../data/raw/20251110_TestScheduleTool.xlsx"

df_tank  = pd.read_excel(file_path, sheet_name="RecipeTank", dtype={"RecipeNumber": str})
df_lid   = pd.read_excel(file_path, sheet_name="RecipeLid",  dtype={"RecipeNumber": str})
df_base  = pd.read_excel(file_path, sheet_name="RecipeBase", dtype={"RecipeNumber": str})

# One big recipe table
df_recipes = pd.concat([df_tank, df_lid, df_base], ignore_index=True)
df_recipes["RecipeNumber"] = df_recipes["RecipeNumber"].astype(int).astype(str).str.zfill(3)
df_recipes = df_recipes.sort_values(by="RecipeNumber").reset_index(drop=True)

df_recipes.head()
```
